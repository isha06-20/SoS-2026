# Week 7: Analysis of Results and Visualization

## Objective

In this week, the focus is on analyzing the performance of the machine learning models developed in the previous weeks and identifying the key factors affecting dropout rates after Class 10 in India.

The analysis includes:

- Comparing the performance of Linear Regression and Decision Tree Regression models.
- Evaluating the impact of hyperparameter tuning techniques such as Grid Search and Random Search.
- Interpreting the feature importance obtained from the Decision Tree model.
- Creating visualizations to summarize the findings.
- Drawing conclusions based on the overall results.

This analysis helps in understanding which factors have the greatest influence on dropout rates and which machine learning approach is most suitable for this dataset.

```python
## 1. Model Performance Summary
# This section summarizes the evaluation metrics of all regression models developed during Weeks 5 and 6. The comparison is based on three regression evaluation metrics:
#- Mean Absolute Error (MAE)
#- Root Mean Squared Error (RMSE)
#- R-squared (R²)
# The summary helps compare the predictive performance of each model before performing a detailed analysis using visualizations.
# Import pandas for creating the summary table
import pandas as pd
# Create a dictionary containing the evaluation metrics
performance = {
    "Model": [
        "Linear Regression",
        "Decision Tree",
        "Decision Tree (Grid Search)",
        "Decision Tree (Random Search)"
    ],

    "MAE": [
        4.12,
        5.38,
        6.547,
        6.540
    ],

    "RMSE": [
        5.40,
        7.53,
        8.672,
        9.030
    ],

    "R² Score": [
        0.024,
        -0.904,
        -1.522,
        -1.735
    ]
}
# Convert the dictionary into a DataFrame
performance_df = pd.DataFrame(performance)
# Display the summary table
performance_df
```

```python
## 2. Comparison of Mean Absolute Error (MAE)
# Mean Absolute Error (MAE) measures the average absolute difference between the actual and predicted dropout rates.
# A lower MAE indicates better prediction accuracy. This visualization compares the MAE values of all regression models developed during the project.
# Import matplotlib for visualization
import matplotlib.pyplot as plt
# Create a bar chart for MAE comparison
plt.figure(figsize=(10,6))
plt.bar(
    performance_df["Model"],
    performance_df["MAE"]
)
# Add chart title and axis labels
plt.title("Comparison of Mean Absolute Error (MAE)")
plt.xlabel("Regression Models")
plt.ylabel("MAE")
# Rotate model names for better readability
plt.xticks(rotation=15)
# Display MAE values on top of each bar
for i, value in enumerate(performance_df["MAE"]):
    plt.text(i, value + 0.08, f"{value:.2f}", ha='center')
# Improve layout
plt.tight_layout()
# Display the graph
plt.show()
```

```python
## 3. Comparison of Root Mean Squared Error (RMSE)
# Root Mean Squared Error (RMSE) measures the square root of the average squared differences between the actual and predicted dropout rates.
# Unlike MAE, RMSE gives greater weight to larger prediction errors. Therefore, a lower RMSE indicates better model performance and fewer large prediction errors.
# Create a bar chart for RMSE comparison
plt.figure(figsize=(8,5))
plt.bar(
    performance_df["Model"],
    performance_df["RMSE"]
)
# Add chart title and axis labels
plt.title("Comparison of Root Mean Squared Error (RMSE)")
plt.xlabel("Regression Models")
plt.ylabel("RMSE")
# Rotate x-axis labels for better readability
plt.xticks(rotation=15)
# Display RMSE values on top of each bar
for i, value in enumerate(performance_df["RMSE"]):
    plt.text(i, value + 0.1, f"{value:.2f}", ha='center')
# Improve layout
plt.tight_layout()
# Display the graph
plt.show()
```

```python
## 4. Comparison of R² Score
# The R² (Coefficient of Determination) measures how well a regression model explains the variation in the target variable.
# - An R² value closer to **1** indicates better model performance.
# - An R² value close to **0** indicates poor explanatory power.
# - A **negative R²** suggests that the model performs worse than simply predicting the mean of the target variable.
# The following graph compares the R² scores of all regression models developed during the project.
# Create a horizontal bar chart for R² Score comparison
plt.figure(figsize=(20, 12))
plt.barh(
    performance_df["Model"],
    performance_df["R² Score"]
)
# Reference line at R² = 0
plt.axvline(x=0, linestyle='--')
# Title and labels
plt.title("Comparison of R² Score")
plt.xlabel("R² Score")
plt.ylabel("Regression Models")
# Display values on each bar
for i, value in enumerate(performance_df["R² Score"]):
    if value >= 0:
        plt.text(value + 0.03, i, f"{value:.3f}", va='center')
    else:
        plt.text(value - 0.25, i, f"{value:.3f}", va='center')
plt.tight_layout()
plt.show()
```

```python
## 5. Feature Importance Analysis
# Feature importance indicates the contribution of each predictor variable in making predictions using the Decision Tree Regression model.
# Features with higher importance values have a greater influence on predicting dropout rates, while features with lower importance values contribute less to the model.
# The following visualization highlights the relative importance of the selected features used in the Decision Tree model.
import pandas as pd
import matplotlib.pyplot as plt
# Placeholder for feature_importance_df (replace with actual data from your model)
feature_importance_data = {
    "Feature": ["Parental Education", "Family Income", "School Type", "Region", "Gender", "Number of Siblings", "Internet Access"],
    "Importance": [0.25, 0.20, 0.15, 0.12, 0.10, 0.09, 0.09]
}
feature_importance_df = pd.DataFrame(feature_importance_data)
# Sort feature importance values in descending order
feature_importance_df = feature_importance_df.sort_values(
    by="Importance",
    ascending=False
)
# Create the plot
plt.figure(figsize=(10,6))
plt.barh(
    feature_importance_df["Feature"],
    feature_importance_df["Importance"]
)
# Display the most important feature at the top
plt.gca().invert_yaxis()
# Add title and labels
plt.title("Feature Importance from Decision Tree Model")
plt.xlabel("Importance Score")
plt.ylabel("Features")
# Improve spacing
plt.tight_layout()
# Display the graph
plt.show()
```

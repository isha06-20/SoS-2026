```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
plt.style.use('default')
pd.set_option('display.max_columns', None)
# load cleaned datasets
gender=pd.read_csv('/content/gender_clean.csv')
trend=pd.read_csv('/content/trend_clean.csv')
level=pd.read_csv('/content/level_clean.csv')
# dataset overview
print("Gender Dataset:")
print(gender.head())
print("Trend Dataset:")
print(trend.head())
print("Level Dataset:")
print(level.head())
```

```python
# datset shapes
print("Gender Dataset Shape:", gender.shape)
print("Trend Dataset Shape:", trend.shape)
print("Level Dataset Shape:", level.shape)
```

```python
# summary stats
display(gender.describe())
display(trend.describe())
display(level.describe())
```

```python
# missing values check
print("Gender Missing Values:")
print(gender.isnull().sum())
print("\nTrend Missing Values:")
print(trend.isnull().sum())
print("\nLevel Missing Values:")
print(level.isnull().sum())
```

```python
print("gender")
print(gender.columns.tolist())
print("trend")
print(trend.columns.tolist())
print("level")
print(level.columns.tolist())
```

```python
# top ten states with highest secondary dropout
top10 = level.sort_values(by='Secondary Drop Out Rate - Overall', ascending=False).head(10)
plt.figure(figsize=(15, 5))
sns.barplot(x='State/UT', y='Secondary Drop Out Rate - Overall', data=top10)
plt.title('Top 10 States with Highest Secondary Dropout')
plt.xlabel('State')
plt.ylabel('Secondary Dropout Rate - Overall (%)')
plt.show()
print(top10[['State/UT', 'Secondary Drop Out Rate - Overall']])
```

```python
# bottom 10 states
bottom10 = level.sort_values(by='Secondary Drop Out Rate - Overall').head(10)
plt.figure(figsize=(20, 5))
sns.barplot(x='State/UT', y='Secondary Drop Out Rate - Overall', data=bottom10)
plt.title('Bottom 10 States with Lowest Secondary Dropout')
plt.xlabel('State')
plt.ylabel('Secondary Dropout Rate - Overall (%)')
plt.show()
print(bottom10[['State/UT', 'Secondary Drop Out Rate - Overall']])
```

```python
# Boys vs girls secondary dropout
comparison = level[['State/UT', 'Secondary Drop Out Rate - Boys ', 'Secondary Drop Out Rate - Girls']]
comparison = comparison.set_index('State/UT')
comparison.plot(kind='bar', figsize=(20, 6))
plt.title('Secondary Dropout Rate: Boys vs Girls')
plt.xlabel('State')
plt.ylabel('Secondary Dropout Rate (%)')
plt.show()
```

```python
# gender gap analysis
level['Gender Gap'] = level['Secondary Drop Out Rate - Boys '] - level['Secondary Drop Out Rate - Girls']
gap = level.sort_values(by='Gender Gap', ascending=False).head(10)
plt.figure(figsize=(15,3))
sns.barplot(x='State/UT', y='Gender Gap', data=gap)
plt.title(" Top 10 State/UT with Gender Gap in Secondary Dropout ")
plt.xticks(rotation=90)
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
plt.style.use('default')
pd.set_option('display.max_columns', None)

# National average by Education level
education_levels ={
    'Primary':
     gender['Primary level (I - V) - Total'].mean(),
    'Upper Primary':
    gender['Upper Primary - Total'].mean(),
    'Elementary':
    gender['Elementary level (I - VIII) - Total'].mean(),
    'Secondary':
    gender['Secondary (IX-X) - Total'].mean(),
    'Higher Secondary':
    gender['Higher Secondary - Total'].mean()

}

plt.figure(figsize=(16, 5))
plt.bar(education_levels.keys(), education_levels.values())
plt.title('National Average by Education Level')
plt.xlabel('Education Level')
plt.ylabel('Dropout Rate (%)')
plt.show()
```

```python
# trend analysis
years =['2017-18', '2018-19', '2019-20']
primary_avg=[
    trend['2017-18 - Primary'].mean(),
    trend['2018-19 - Primary'].mean(),
    trend['2019-20 - Primary'].mean()

]
secondary_avg=[
    trend['2017-18 - Secondary'].mean(),
    trend['2018-19 - Secondary'].mean(),
    trend['2019-20 - Secondary'].mean()

]
plt.figure(figsize=(16, 5))
plt.plot(years, primary_avg, marker='o', label='Primary')
plt.plot(years, secondary_avg, marker='o', label='Secondary')
plt.title('Trend Analysis: Average Dropout Rates over years')
plt.xlabel('Year')
plt.ylabel('Dropout Rate (%)')
plt.legend()
plt.show()
```

```python
# states above national average
national_avg =( level['Secondary Drop Out Rate - Overall'].mean()
)
states_above_avg = level[level['Secondary Drop Out Rate - Overall'] > national_avg]
states_above_avg = states_above_avg.sort_values(by='Secondary Drop Out Rate - Overall', ascending=False)
print(" National Average:" , national_avg)
print("States Above National Average:")
print(states_above_avg[['State/UT', 'Secondary Drop Out Rate - Overall']])
display(states_above_avg['State/UT'])
display(states_above_avg['Secondary Drop Out Rate - Overall'])
```

```python
# correlation heatmap
numeric_data = level.select_dtypes(include=[ 'float64', 'int64'])
plt.figure(figsize=(15, 6))
sns.heatmap(numeric_data.corr(), annot=True, cmap='coolwarm')
plt.title('Correlation Heatmap')
plt.show()
```

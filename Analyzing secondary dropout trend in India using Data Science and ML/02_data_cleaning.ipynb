```python
# import libraries
import pandas as pd
import numpy as np
pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', None)
```

```python
import pandas as pd
# loading and verifying data
df_gender = pd.read_csv('gender.csv')
df_trend = pd.read_csv('trend.csv')
df_level = pd.read_csv('level.csv')

df_gender.head()
df_trend.head()
df_level.head()
```

```python
#create copies
df_gender_copy = df_gender.copy()
df_trend_copy = df_trend.copy()
df_level_copy = df_level.copy()
```

```python
# inspection of dataset
print("Gender Dataset Shape:" , df_gender_copy.shape)
print("Trend Dataset Shape:" , df_trend_copy.shape)
print("Level Dataset Shape:" , df_level_copy.shape)
print(df_gender_copy.columns)
print(df_trend_copy.columns)
print(df_level_copy.columns)
```

```python
# remove serial number coloumns
df_gender_copy.drop(['Sl. No.'], axis=1, inplace=True, errors='ignore')
df_trend_copy.drop(['Sl. No.'], axis=1, inplace=True, errors='ignore')
df_level_copy.drop(['Sl. No.'], axis=1, inplace=True, errors='ignore')
print(" serial number coloumns removed")
# gender and trend doesn't has sl no
```

```python
for df in [df_gender_copy, df_trend_copy, df_level_copy]:
  state_col=df.columns[0]
  df.drop(
      df[df[state_col].astype(str)
      .str.strip()
      .str.lower() == 'india']. index, inplace=True)
  print(" India row removed")
```

```python
#reset index
df_gender_copy.reset_index(drop=True, inplace=True)
df_trend_copy.reset_index(drop=True, inplace=True)
df_level_copy.reset_index(drop=True, inplace=True)
```

```python
#standardise state names
df_gender_copy['State/UT'] = df_gender_copy['State/UT'].str.title()
df_trend_copy['State/UT'] = df_trend_copy['State/UT'].str.title()
df_level_copy['State/UT']= df_level_copy['State/UT'].str.title()
```

```python
# missing values check
print("=======GENDER DATASET=======")
print(df_gender_copy.isnull().sum())
print("=======TREND DATASET=======")
print(df_trend_copy.isnull().sum())
print("=======LEVEL DATASET=======")
print(df_level_copy.isnull().sum())
```

```python
# converet numeric column
for df in[df_gender_copy, df_trend_copy, df_level_copy]:
  state_col = df.columns[0]
  numeric_cols=[col for col in df.columns if col!=state_col]
  for col in numeric_cols:
    df[col]=pd.to_numeric(df[col], errors='coerce')
    #verify data types
    print(df_gender_copy.info())
    print(df_trend_copy.info())
    print(df_level_copy.info())
```

```python
# descriptive stats
print("=======GENDER DATASET=======")
print(df_gender_copy.describe())
print("=======TREND DATASET=======")
print(df_trend_copy.describe())
print("=======LEVEL DATASET=======")
print(df_level_copy.describe())
```

```python
#duplicate check
print(" gender duplicates:",df_gender_copy.duplicated().sum())
print(" trend duplicates:",df_trend_copy.duplicated().sum())
print(" level duplicates:",df_level_copy.duplicated().sum())
```

```python
# state coverage comparision
states_gender= set(df_gender_copy.iloc[:,0])
states_trend= set(df_trend_copy.iloc[:,0])
states_level= set(df_level_copy.iloc[:,0])
print( "States in Gender:",len(states_gender))
print( "States in Trend:",len(states_trend))
print( "States in Level:",len(states_level))
```

```python
#states missing across datasets
print (" Present in gender but not in trend ")
print(states_gender-states_trend)
print (" Present in trend but not in level ")
print(states_trend-states_level)
print (" Present in level but not in gender ")
print(states_level-states_gender)
```

```python
# check suspicious zero value
for col in df_gender_copy.columns[1:]:
  zeros = (df_gender_copy[col]==0).sum()
  print(f"{col}: {zeros}")
```

```python
# save cleaned files
df_gender_copy.to_csv('gender_cleaned.csv', index=False)
df_trend_copy.to_csv('trend_cleaned.csv', index=False)
df_level_copy.to_csv('level_cleaned.csv', index=False)
print(" All cleaned files saved successfully")
```
